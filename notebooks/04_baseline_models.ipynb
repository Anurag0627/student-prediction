# 04 — Baseline Models (Experiment A)

Train all models **without** any imbalance handling to establish the baseline. This is the reference every later experiment is compared to.

```python
import sys
from pathlib import Path

# Make the project root importable (so `import src...` works).
ROOT = Path.cwd()
if (ROOT / 'src').exists() is False and (ROOT.parent / 'src').exists():
    ROOT = ROOT.parent
sys.path.insert(0, str(ROOT))

import pandas as pd
import matplotlib.pyplot as plt

from src.utils import config
config.ensure_directories()
```

```python
from src.data import load_data, preprocessing
from src.models import baseline, train

df = load_data.load_dataset()
X, y = preprocessing.separate_features_target(df)
X_train, X_test, y_train, y_test = preprocessing.split_data(X, y)
```

## Run the baseline experiment for every model

```python
results = []
for name, estimator in baseline.get_models().items():
    results.append(train.run_experiment(
        model_name=name, estimator=estimator, strategy='none',
        X_train=X_train, X_test=X_test, y_train=y_train, y_test=y_test,
    ))
results_df = train.results_to_frame(results)
results_df
```

## Save baseline metrics

```python
out = config.METRICS_DIR / 'baseline_results.csv'
results_df.to_csv(out, index=False)
print('saved ->', out)
```

### Notes

Record observations only — do **not** conclude which technique is best; no imbalance handling has been applied yet.
