# 03 — Preprocessing

Build the **leakage-safe** preprocessing pipeline.

**Rule:** split first, then fit preprocessing on the training set only.

```python
import sys
from pathlib import Path

# Make the project root importable (so `import src...` works).
ROOT = Path.cwd()
if (ROOT / 'src').exists() is False and (ROOT.parent / 'src').exists():
    ROOT = ROOT.parent
sys.path.insert(0, str(ROOT))

import pandas as pd
import matplotlib.pyplot as plt

from src.utils import config
config.ensure_directories()
```

```python
from src.data import load_data, preprocessing

df = load_data.load_dataset()
X, y = preprocessing.separate_features_target(df)
X_train, X_test, y_train, y_test = preprocessing.split_data(X, y)
print('train:', X_train.shape, '| test:', X_test.shape)
```

## Feature typing

```python
num_cols = preprocessing.identify_numerical_features(X_train)
cat_cols = preprocessing.identify_categorical_features(X_train)
print(len(num_cols), 'numeric |', len(cat_cols), 'categorical')
```

## Build the ColumnTransformer (unfitted)

It is fit **inside** a pipeline on the training data only — never on the full dataset.

```python
preprocessor = preprocessing.build_preprocessor(num_cols, cat_cols)
preprocessor
```

## Sanity check: fit on train, transform both

Demonstrates the transformer works; in experiments this is handled automatically by the pipeline.

```python
fitted = preprocessor.fit(X_train)
print('train transformed:', fitted.transform(X_train).shape)
print('test transformed :', fitted.transform(X_test).shape)
```
