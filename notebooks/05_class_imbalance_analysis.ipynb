## Why accuracy misleads

A model that always predicts the majority class (`Graduate`) achieves accuracy equal to the majority share, yet has **zero recall** on `Dropout`. Compute that naive baseline for context:

```python
majority_share = imbalance.class_percentages(y).max()
print(f'Always-Graduate accuracy would be ~{majority_share:.2f}% '
      'with 0% Dropout recall.')
```
